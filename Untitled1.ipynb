```python
import numpy as np

temperatures = np.array([20, 22, 23, 25, 27])

print("Temperature:", temperatures)
print("Mean:", np.mean(temperatures))
print("Maximun:", np.max(temperatures))
print("Minimun:", np.min(temperatures))
```

```python
import pandas as pd

data = {
    "Espece": ["Rose", "Tulipe", "Marguerite"],
    "Temperature": [20, 25, 22]
}
df = pd.DataFrame(data)
print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=df, x="Espece", y="Temperature")
plt.show()
```
